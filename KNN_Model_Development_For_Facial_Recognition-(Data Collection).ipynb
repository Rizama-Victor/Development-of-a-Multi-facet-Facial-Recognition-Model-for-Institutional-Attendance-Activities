### Importation of Necessary Libraries

```python
# -- imports OpenCV library for the image capturing process
import cv2 

# -- imports the operating system library for handling file paths and directories
import os 

# -- imports numpy library for numerical operations
import numpy as np 

# -- imports the PIL library for image manipulation
from PIL import Image 
```

#### Function to Generate Dataset

```python
def generate_dataset():
    # Loads the Haar cascade face detector
    face_classifier = cv2.CascadeClassifier("haarcascade_frontalface_default.xml")
    
    # Function to detect and crop the face from a frame
    def face_cropped(img):
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY) # converts image from BGR to grayscale
        faces = face_classifier.detectMultiScale(gray, 1.3, 5) # detects the faces in an image
        # scaling factor = 1.3 reduces the image size at each scale to detect smaller faces
        # minimum neighbor = 5 is the minimum number of neighbors each rectangle should have to retain it
        
        if len(faces) == 0:
            return None # returns None if no faces is detected
        
        # Iterates through the detected face
        for (x, y, w, h) in faces:
            padding = 30
            # x = max(0, x - padding)
            # y = max(0, y - padding)
            # w = w + 2 * padding
            # h = h + 2 * padding

            cropped_face = img[y:y+h, x:x+w] # crops the detected face region
            return cropped_face # returns the cropped face

    # Prompts user to enter their ID (which will be used for file naming)
    user_id = input("Enter user ID (e.g., 1): ")
    
    # Creates the data directory if it doesn't exist.
    if not os.path.exists("data"):
        os.makedirs("data")

    # Starts the webcam capture (change index to 0 or 1 depending on your device)
    cap = cv2.VideoCapture(0)
    img_id = 0 # counter for the number of captured images

    while True:
        ret, frame = cap.read() # reads a single frame
        if not ret:
            continue # skips the current iteration if not there's an error in reading the file

        face = face_cropped(frame) # detects and crops the face from the current frame

        # Processes and saves the face if detected
        if face is not None:
            img_id += 1 # increments the counter

            # Resizes face to a 400 x 400 pixels for uniformity
            face = cv2.resize(face, (400, 400))
            face = cv2.cvtColor(face, cv2.COLOR_BGR2GRAY) # converts face image to grayscale

            file_name_path = f"data/user.{user_id}.{img_id}.jpg" # constructs a unique file name for each image
            cv2.imwrite(file_name_path, face) # saves the processed face image to the dataset folder

            # Displays the captured face with image count overlay
            cv2.putText(face, str(img_id), (50, 50), cv2.FONT_HERSHEY_COMPLEX, 1, (0, 255, 0), 2)
            cv2.imshow("Cropped Face", face)

        # Stop after collecting 200 images or pressing Enter (key code 13)
        if cv2.waitKey(1) == 13 or img_id == 30:
            break

    # Releases webcam and closes all OpenCV windows
    cap.release()
    cv2.destroyAllWindows()
    print("Collecting samples is completed....")
```

```python

generate_dataset() # calls the generate dataset function to perform the data collection procedure
```
